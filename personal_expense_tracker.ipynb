```python
import matplotlib.pyplot as plt

print("====================================")
print("       PERSONAL EXPENSE TRACKER")
print("====================================")

# Enter monthly budget
budget = float(input("Enter your monthly budget (Rs.): "))

expenses = []

# Enter number of expenses
n = int(input("\nHow many expenses do you want to enter? "))

# Collect expense details
for i in range(n):

    print("\nExpense Number:", i + 1)

    name = input("Enter expense name: ")
    category = input("Enter category (Food/Travel/Shopping/Bills/Other): ")

    while True:
        amount = float(input("Enter amount (Rs.): "))

        if amount >= 0:
            break
        else:
            print("Please enter a valid amount.")

    expenses.append({
        "name": name,
        "category": category.title(),
        "amount": amount
    })


# Display expenses
print("\n========== EXPENSE DETAILS ==========")

for e in expenses:
    print(
        "Name:", e["name"],
        "| Category:", e["category"],
        "| Amount: Rs.", e["amount"]
    )


# Calculate total expenses
total = sum(e["amount"] for e in expenses)
balance = budget - total

print("\n========== FINANCIAL SUMMARY ==========")

print("Monthly Budget: Rs.", budget)
print("Total Expenses: Rs.", total)
print("Remaining Balance: Rs.", balance)

if balance < 0:
    print("Warning! You have exceeded your budget.")
else:
    print("You are within your budget.")


# Category-wise analysis
categories = {}

for e in expenses:
    category = e["category"]
    categories[category] = categories.get(category, 0) + e["amount"]

print("\n========== CATEGORY ANALYSIS ==========")

for category, amount in categories.items():
    print(category, ": Rs.", amount)


# Find highest expense
if expenses:
    highest = max(expenses, key=lambda x: x["amount"])
    print("\nHighest Expense:", highest["name"])
    print("Amount: Rs.", highest["amount"])


# Bar Chart
plt.figure(figsize=(8, 5))
plt.bar(categories.keys(), categories.values())
plt.title("Category Wise Expense Analysis")
plt.xlabel("Expense Categories")
plt.ylabel("Amount (Rs.)")
plt.show()


# Pie Chart
plt.figure(figsize=(7, 7))
plt.pie(
    categories.values(),
    labels=categories.keys(),
    autopct="%1.1f%%"
)
plt.title("Expense Distribution")
plt.show()


# Budget Analysis Chart
plt.figure(figsize=(7, 5))
plt.bar(
    ["Monthly Budget", "Total Expenses", "Remaining Balance"],
    [budget, total, balance]
)
plt.title("Monthly Budget Analysis")
plt.ylabel("Amount (Rs.)")
plt.xticks(rotation=15)
plt.show()

print("\nProject Completed Successfully!")
```
